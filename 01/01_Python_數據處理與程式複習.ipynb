## 第一節｜Python 基礎

學習順序：**變數與資料型態 → list → dict → 巢狀結構 → if → for → 函式**。

記住：`[]` 可建立 list；`{}` 可建立 dict；從 list 取值使用數字索引，從 dict 取值使用 key。

### 練習 1-1｜變數與基本資料型態

`int` 表示整數、`float` 表示小數、`str` 表示文字、`bool` 表示真假值。`type()` 可以檢查型態。

**參考程式（先看懂，再輸入下方空白 Code cell）：**

```python
hours = 3
time = 3.5
group = "宣傳"
ready = hours >= 3
print(type(hours), type(time))
print(type(group), type(ready))
```

**輪到你：** 把 `hours` 改成 `2`，預測 `ready` 會變成什麼，再加入 `print(ready)` 檢查。

### 練習 1-2｜變數可以參與運算

給數字取名字後，只要修改輸入值，就可以讓運算結果一起改變。這裡假設每小時可完成兩項任務。

**參考程式（先看懂，再輸入下方空白 Code cell）：**

```python
hours = 3
tasks = hours * 2
print(tasks)
```

**輪到你：** 將 `hours` 改為 `4`，再試著修改每小時的任務數。

### 練習 1-3｜list：用 [] 保存多筆資料

`list` 依照順序存放元素。**第一筆索引為 0**，`-1` 代表最後一筆。

**參考程式（先看懂，再輸入下方空白 Code cell）：**

```python
hours = [2, 4, 3, 5, 1, 3, 4, 2]
print(hours[0])
print(hours[3])
print(hours[-1])
```

**輪到你：** 找出第 2 筆和第 5 筆的數字，注意索引要比第幾筆少 1。

### 練習 1-4｜dict：用 {} 保存鍵和值

`dict` 使用 `key: value` 的方式存資料，不需要記住資料放在第幾個位置。

**參考程式（先看懂，再輸入下方空白 Code cell）：**

```python
data = {"id": "A", "group": "宣傳", "hours": 2, "tasks": 3}
print(data["group"])
print(data["hours"])
```

**輪到你：** 增加一個 `"name": "Amy"` 的欄位，再印出 `data["name"]`。

### 練習 1-5｜巢狀 dict 與 list：一層一層取值

一個 dict 的 value 還可以是另一個 dict，裡面也可以再放 list。取值時，從外層依序寫出每個 key 或索引。

**參考程式（先看懂，再輸入下方空白 Code cell）：**

```python
data = {
    "name": "Amy",
    "info": {"hours": 2, "days": ["Mon", "Wed"]}
}
print(data["info"]["hours"])
print(data["info"]["days"][0])
```

**輪到你：** 改成取出 `"Wed"`。想想最後一個 `[]` 應填入哪個索引。

### 練習 1-6｜if／else：根據條件做決定

使用 `>=` 表示「大於或等於」。`if` 和 `else` 下方都必須縮排。

**參考程式（先看懂，再輸入下方空白 Code cell）：**

```python
hours = 3
if hours >= 3:
    print("達標")
else:
    print("未達標")
```

**輪到你：** 將 `hours` 改為 `2`，再改為 `5`，比較執行結果。

### 練習 1-7｜for 迴圈：依序處理 list

迴圈每次從 `tasks` 取出一筆，累加到 `total`，就不需要重複寫四次加法。

**參考程式（先看懂，再輸入下方空白 Code cell）：**

```python
tasks = [3, 5, 4, 8]
total = 0
for task in tasks:
    total += task
print(total)
```

**輪到你：** 在 list 最後再加入 `2`，預測總和，然後執行確認。

### 練習 1-8｜函式：用 def 包住可重複的步驟

`def` 建立函式，括號內是輸入參數，`return` 傳回結果。

**參考程式（先看懂，再輸入下方空白 Code cell）：**

```python
def get_tasks(hours):
    return hours * 2

print(get_tasks(3))
print(get_tasks(5))
```

**輪到你：** 把每小時完成的任務數改成 `3`，觀察兩次函式呼叫的結果。

**第一節自我檢查**：能說出 list `[]`、dict `{}` 的差異；能用 `if` 判斷、用 `for` 加總、用 `def` 定義函式，就可以開始下一節。

## 第二節｜NumPy 陣列

`NumPy` 用於處理整批數字。這一節先比較 list 和 array，再練習**索引、切片、二維陣列、向量化、篩選與統計**。

> 以下範例依順序執行，第一次練習會匯入 `numpy`，後面就可以使用 `np`。

### 練習 2-1｜Python list 和 NumPy array 的乘法不同

Python list 乘以 2 是**重複內容**；NumPy array 乘以 2 則是**每個元素各乘以 2**。

**參考程式（先看懂，再輸入下方空白 Code cell）：**

```python
import numpy as np
items = [2, 4, 3]
a = np.array(items)
print(items * 2)
print(a * 2)
```

**輪到你：** 把乘數由 `2` 改成 `3`，比較兩種輸出的長度和數值。

### 練習 2-2｜一維陣列：索引與切片

`a[start:stop]` 會從 `start` 取到 `stop` 的前一個索引，**不包含 stop**。

**參考程式（先看懂，再輸入下方空白 Code cell）：**

```python
hours = np.array([2, 4, 3, 5, 1, 3, 4, 2])
print(hours.shape)
print(hours[0])
print(hours[1:4])
```

**輪到你：** 試著取得索引 `2`、`3`、`4` 的三筆資料。

### 練習 2-3｜二維陣列：先列、再欄

二維陣列像表格，`a.shape` 表示（列數、欄數）；`a[row, col]` 先寫列索引再寫欄索引。

**參考程式（先看懂，再輸入下方空白 Code cell）：**

```python
a = np.array([
    [2, 4, 3, 5],
    [1, 3, 4, 2],
    [3, 5, 2, 4]
])
print(a.shape)
print(a[1, 2])
```

**輪到你：** 找出第 3 列、第 1 欄的數值；別忘記索引從 0 開始。

### 練習 2-4｜向量化：整個陣列一起運算

不用寫迴圈，也能讓每筆時數增加 2。兩個長度相同的陣列，也可以對應位置相加。

**參考程式（先看懂，再輸入下方空白 Code cell）：**

```python
new = hours + 2
x = np.array([1, 2, 3])
y = np.array([4, 5, 6])
print(new)
print(x + y)
```

**輪到你：** 把 `hours + 2` 改成 `hours * 2`，看看每筆時數如何變化。

### 練習 2-5｜布林篩選：只留下符合條件的值

`hours >= 3` 會對每筆數字判斷 True 或 False，當作遮罩留下符合門檻的資料。

**參考程式（先看懂，再輸入下方空白 Code cell）：**

```python
good = hours[hours >= 3]
print(good)
print(len(good))
```

**輪到你：** 把門檻 `3` 改成 `4`，觀察剩下幾筆資料。

### 練習 2-6｜快速統計：sum、mean、min、max

四個統計方法分別回答「總和、平均、最小、最大」。

**參考程式（先看懂，再輸入下方空白 Code cell）：**

```python
print(hours.sum())
print(hours.mean())
print(hours.min())
print(hours.max())
```

**輪到你：** 說出平均時數與最多時數分別是多少。

### 練習 2-7｜axis：每欄平均與每列平均

二維陣列的 `axis=0` 會留下**每一欄**的平均，`axis=1` 會留下**每一列**的平均。這裡沿用前面的 3 列 × 4 欄的 `a`。

**參考程式（先看懂，再輸入下方空白 Code cell）：**

```python
print(a.mean(axis=0))
print(a.mean(axis=1))
```

**輪到你：** 先預測兩次輸出各有幾個數字，再執行檢查。

**第二節自我檢查**：可以用 array 進行整批加減乘除、利用條件選資料，並知道 `axis=0` 和 `axis=1` 的結果長度不同。

## 第三節｜Pandas 與 Matplotlib

這節使用同一份 **`activity.csv`**（8 筆教學模擬活動紀錄）完成資料讀取、整理、分組，再畫出 **bar／hist／pie** 三種圖表。

### 先確認資料檔案！

- **Google Colab**：先到左側的「檔案」區上傳 `activity.csv`；**重新連線或重啟環境後，通常要重新上傳**。
- **VS Code / Jupyter**：將 `activity.csv` 放在目前 Notebook 執行的工作資料夾。
- 如果出現 `FileNotFoundError`，通常代表程式找不到這份 CSV，請先確認檔案位置，再執行讀取程式。

檔案欄位為 `id`（代號）、`group`（小組）、`hours`（準備時數）、`tasks`（完成任務）。**F 的 hours 是空白，表示缺漏，不代表 0。**

| id | group | hours | tasks |
|:--:|:--:|--:|--:|
| A | 宣傳 | 2 | 3 |
| B | 場佈 | 4 | 5 |
| C | 宣傳 | 3 | 4 |
| D | 接待 | 5 | 8 |
| E | 場佈 | 1 | 2 |
| F | 接待 | 缺漏 | 5 |
| G | 宣傳 | 4 | 6 |
| H | 場佈 | 2 | 3 |

> Matplotlib 圖上的組別標籤使用簡單英文（Promo／Setup／Welcome），依序對應**宣傳／場佈／接待**，避免有些執行環境缺少中文字型而顯示方框。

### 練習 3-1｜讀取 CSV，認識 DataFrame

`pd.read_csv()` 將 CSV 讀入 `df`；`head()` 檢查前幾筆，`shape` 查看（列數、欄數）。

**參考程式（先看懂，再輸入下方空白 Code cell）：**

```python
import pandas as pd
df = pd.read_csv("activity.csv")
print(df.head())
print(df.shape)
```

**輪到你：** 確認資料是不是 **8 列、4 欄**；如果找不到檔案，先上傳 `activity.csv`。

### 練習 3-2｜選取想看的欄位

使用 `df["hours"]` 選**一欄**；使用 `df[["group", "hours"]]` 選**兩欄**。注意雙中括號。

**參考程式（先看懂，再輸入下方空白 Code cell）：**

```python
print(df["hours"])
print(df[["group", "hours"]])
```

**輪到你：** 改成只顯示 `id` 和 `tasks` 兩欄。

### 練習 3-3｜條件篩選：找出任務數達標者

用比較條件從 DataFrame 留下符合要求的列。

**參考程式（先看懂，再輸入下方空白 Code cell）：**

```python
good = df[df["tasks"] >= 5]
print(good)
```

**輪到你：** 把門檻由 `5` 改成 `6`，看看還有哪幾筆紀錄。

### 練習 3-4｜缺失值：先看清楚，再決定怎麼處理

`isna()` 查缺漏；`mean()` 算已知時數的平均；`fillna()` 可把缺漏改用平均補值。這裡建立 `new`，保留原始的 `df`。

**參考程式（先看懂，再輸入下方空白 Code cell）：**

```python
print(df["hours"].isna())
print(df["hours"].mean())
new = df.copy()
new["hours"] = new["hours"].fillna(new["hours"].mean())
print(new[["id", "hours"]])
```

**輪到你：** 找出哪一筆原本缺漏；再看看 `new` 的時數是否都能計算。

### 練習 3-5｜groupby：依小組加總與算平均

`groupby()` 會把相同小組的資料放在一起。`sum()` 算任務總數；`mean()` 算平均時數（忽略缺漏）。

**參考程式（先看懂，再輸入下方空白 Code cell）：**

```python
data = df.groupby("group", sort=False)
print(data["tasks"].sum())
print(data["hours"].mean())
```

**輪到你：** 比較「宣傳」、「場佈」、「接待」三組的任務總數是否相同。

### 練習 3-6｜Matplotlib bar：比較各小組任務數

先由 Pandas 將 `tasks` 依組加總，**再把結果交給 Matplotlib 的 `plt.bar()`**。

**參考程式（先看懂，再輸入下方空白 Code cell）：**

```python
import matplotlib.pyplot as plt
data = df.groupby("group", sort=False)["tasks"].sum()
names = ["Promo", "Setup", "Welcome"]
plt.bar(names, data)
plt.ylabel("Tasks")
plt.show()
```

**輪到你：** 把 `"tasks"` 改為 `"hours"`，並把 `sum()` 改為 `mean()`，比較各組平均時數。

### 練習 3-7｜Matplotlib hist：準備時數的分布

直方圖的橫軸是**數值區間**，不是人名或小組。使用 `dropna()` 排除 F 的缺漏時數，再決定分幾組 `bins`。

**參考程式（先看懂，再輸入下方空白 Code cell）：**

```python
data = df["hours"].dropna()
plt.hist(data, bins=5, edgecolor="white")
plt.xlabel("Hours")
plt.ylabel("Count")
plt.show()
```

**輪到你：** 試試 `bins=3` 和 `bins=6`；資料有沒有變？圖形為什麼不同？

### 練習 3-8｜Matplotlib pie：各小組紀錄占比

`groupby().size()` 計算各組的紀錄筆數。`plt.pie()` 用整個圓表示 100%，顯示各組占多少比例。

**參考程式（先看懂，再輸入下方空白 Code cell）：**

```python
data = df.groupby("group", sort=False).size()
names = ["Promo", "Setup", "Welcome"]
plt.pie(data, labels=names, autopct="%1.1f%%")
plt.show()
```

**輪到你：** 將 `size()` 改成 `["tasks"].sum()`，比較「紀錄筆數占比」和「任務總數占比」的差異。

### 練習 3-9｜整合挑戰：清理、篩選、存成新 CSV

最後把資料清理、條件篩選、輸出檔案串在一起。這裡採**刪除時數缺漏的列**，不是用 0 代替。

**參考程式（先看懂，再輸入下方空白 Code cell）：**

```python
new = df.dropna(subset=["hours"])
good = new[new["tasks"] >= 5]
print(good)
good.to_csv("activity_cleaned.csv", index=False)
```

**輪到你：** 將門檻改成 `4`，重新執行；觀察新產生的 `activity_cleaned.csv`。
