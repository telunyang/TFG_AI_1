## 第一節｜認識像素、RGB／BGR 與影像陣列

依序學習：**讀照片 → 顯示 → shape → 像素位置 → 通道 → 修改及儲存**。先理解影像其實是數值陣列，不需要背複雜的數學公式。

### 練習 1-1｜用 OpenCV 讀取照片

`cv2.imread()` 把圖片讀成可處理的陣列。這張範例照片的寬為 **480**、高為 **320**。若出現 `None`，通常表示檔案尚未上傳或檔名不正確。

**參考程式（先閱讀，再輸入下方空白 Code cell）：**

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("photo.jpg")
print(img.shape)
```

**輪到你：** 確認輸出是否為 `(320, 480, 3)`。如果看到錯誤，檢查 `photo.jpg` 是否在目前工作資料夾。

### 練習 1-2｜正確顯示彩色照片

OpenCV 的 `img` 是 **BGR** 順序，而 `plt.imshow()` 預期 **RGB**。先用 `cv2.cvtColor()` 轉換，照片顏色才會正確。

**參考程式（先閱讀，再輸入下方空白 Code cell）：**

```python
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(rgb)
plt.axis("off")
plt.show()
```

**輪到你：** 把 `plt.axis("off")` 改為 `plt.axis("on")`，觀察圖片周圍的座標。

### 練習 1-3｜影像尺寸：先高、再寬

彩色圖片的 `shape` 是 `(高度, 寬度, 通道數)`。總像素數等於 **高 × 寬**。

**參考程式（先閱讀，再輸入下方空白 Code cell）：**

```python
height, width, channels = img.shape
print("高度：", height)
print("寬度：", width)
print("通道數：", channels)
print("像素總數：", height * width)
```

**輪到你：** 試著計算一張寬 80、高 60 的照片有多少像素。

### 練習 1-4｜讀取一個像素的 BGR 數值

影像座標常寫成 `(x, y)`，但陣列使用 `img[y, x]`，**先列 y，再欄 x**。彩色像素會取得三個數值。

**參考程式（先閱讀，再輸入下方空白 Code cell）：**

```python
color = img[120, 240]
print("BGR：", color)
print("藍色通道：", color[0])
```

**輪到你：** 改成 `img[80, 100]`，比較兩個位置的 BGR 數值是否一樣。

### 練習 1-5｜比較 BGR 和 RGB 的顯示差異

同一張照片如果把 BGR 當 RGB 畫出來，紅色與藍色可能對調。用兩張並排圖片比較差異。

**參考程式（先閱讀，再輸入下方空白 Code cell）：**

```python
plt.figure(figsize=(9, 3))
plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title("BGR 當成 RGB")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.imshow(rgb)
plt.title("正確 RGB")
plt.axis("off")
plt.show()
```

**輪到你：** 觀察照片中紅色杯盤的顏色是否在左、右兩張不同。

### 練習 1-6｜觀察單一顏色通道

一張彩色影像有三個通道，可以用 `cv2.split()` 拆開。在單一通道圖中，**越亮代表該通道數值越高**。

**參考程式（先閱讀，再輸入下方空白 Code cell）：**

```python
blue, green, red = cv2.split(img)
print(red.shape)
plt.imshow(red, cmap="gray", vmin=0, vmax=255)
plt.title("R 紅色通道")
plt.axis("off")
plt.show()
```

**輪到你：** 把 `red` 改為 `blue` 或 `green`，比較杯盤位置的亮度。

### 練習 1-7｜修改部分像素，讓照片出現色塊

`[y1:y2, x1:x2]` 可以選取一塊矩形區域。將該區設為 `(255, 0, 0)`，在 OpenCV 的 **BGR** 中代表藍色。用 `.copy()` 避免直接改變原始 `img`。

**參考程式（先閱讀，再輸入下方空白 Code cell）：**

```python
new = img.copy()
new[40:130, 50:190] = (255, 0, 0)
plt.imshow(cv2.cvtColor(new, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

**輪到你：** 把 `(255, 0, 0)` 改成 `(0, 0, 255)`，觀察色塊變成什麼顏色。

### 練習 1-8｜儲存處理後的影像

用 `cv2.imwrite()` 把陣列儲存成圖片檔；之後可以再用 `cv2.imread()` 讀回來。

**參考程式（先閱讀，再輸入下方空白 Code cell）：**

```python
cv2.imwrite("new.jpg", new)
check = cv2.imread("new.jpg")
print(check.shape)
plt.imshow(cv2.cvtColor(check, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

**輪到你：** 將檔名改成 `my_photo.png` 再執行，看看工作資料夾中是否多了一個檔案。

## 第二節｜縮放、裁切、灰階、二值化與模糊

以下範例沿用第一節讀取的杯盤照片 `img`。這一節每次都比較處理前後，而不是只記住函式名稱。

### 練習 2-1｜resize：修改照片的尺寸

`cv2.resize()` 的尺寸參數是 `(寬, 高)`，與 `img.shape` 的前兩個維度順序不同。

**參考程式（先閱讀，再輸入下方空白 Code cell）：**

```python
small = cv2.resize(img, (160, 107))
print("原始：", img.shape)
print("縮小：", small.shape)
plt.imshow(cv2.cvtColor(small, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

**輪到你：** 將尺寸改為 `(80, 53)`，比較照片的細節與 `shape`。

### 練習 2-2｜比較最近鄰與線性插值

先把照片縮小，再放回 480×320。**最近鄰**容易看到像素格子；**線性插值**通常比較平滑，但都無法憑空還原真實細節。

**參考程式（先閱讀，再輸入下方空白 Code cell）：**

```python
tiny = cv2.resize(img, (80, 53))
near = cv2.resize(tiny, (480, 320), interpolation=cv2.INTER_NEAREST)
linear = cv2.resize(tiny, (480, 320), interpolation=cv2.INTER_LINEAR)
plt.figure(figsize=(9, 3))
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(near, cv2.COLOR_BGR2RGB))
plt.title("最近鄰")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(linear, cv2.COLOR_BGR2RGB))
plt.title("線性插值")
plt.axis("off")
plt.show()
```

**輪到你：** 改用 `(40, 27)` 先縮得更小，看看兩種放大結果的差異是否更明顯。

### 練習 2-3｜crop：裁切照片的一部分

裁切使用 NumPy 切片：`img[上:下, 左:右]`。此處先取 y=35～229，再取 x=90～339。

**參考程式（先閱讀，再輸入下方空白 Code cell）：**

```python
crop = img[35:230, 90:340]
print(crop.shape)
plt.imshow(cv2.cvtColor(crop, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

**輪到你：** 將 `90:340` 改成 `150:400`，觀察裁切範圍向哪個方向移動。

### 練習 2-4｜將彩色照片轉成灰階

灰階主要保留**亮暗**資訊；一般灰階影像由三維 `(H,W,3)` 變成二維 `(H,W)`。**灰階不是只剩純黑與純白**。

**參考程式（先閱讀，再輸入下方空白 Code cell）：**

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
print(gray.shape)
plt.imshow(gray, cmap="gray", vmin=0, vmax=255)
plt.axis("off")
plt.show()
```

**輪到你：** 印出 `gray[120, 240]`，比較灰階值與原彩色 `img[120, 240]` 的差異。

### 練習 2-5｜threshold：設定黑白門檻

`cv2.THRESH_BINARY` 會將**大於門檻**的灰階值變成 255（白），其餘變成 0（黑）。`_` 用來接不需要的回傳值。

**參考程式（先閱讀，再輸入下方空白 Code cell）：**

```python
_, binary = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)
plt.imshow(binary, cmap="gray", vmin=0, vmax=255)
plt.axis("off")
plt.show()
```

**輪到你：** 將門檻從 `127` 改成 `90`，再改成 `180`，觀察白色區域變多還是變少。

### 練習 2-6｜Gaussian Blur：模糊照片

高斯模糊會參考周圍像素，減弱小範圍細節與雜訊。模糊核心大小常用 `(3,3)`、`(5,5)` 等**正的奇數尺寸**。

**參考程式（先閱讀，再輸入下方空白 Code cell）：**

```python
blur = cv2.GaussianBlur(img, (5, 5), 0)
plt.imshow(cv2.cvtColor(blur, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

**輪到你：** 把核心改成 `(15, 15)`，比較圖片哪些細節變得不清楚。

### 練習 2-7｜串起處理流程：灰階 → 模糊 → 二值化

資料處理的**先後順序**會改變結果。這裡先保留亮暗，再平滑細節，最後用門檻分成黑白。

**參考程式（先閱讀，再輸入下方空白 Code cell）：**

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
smooth = cv2.GaussianBlur(gray, (5, 5), 0)
_, binary = cv2.threshold(smooth, 127, 255, cv2.THRESH_BINARY)
plt.imshow(binary, cmap="gray", vmin=0, vmax=255)
plt.axis("off")
plt.show()
```

**輪到你：** 將模糊核心改成 `(9, 9)`，觀察最後的黑白結果有什麼不同。

### 練習 2-8｜一次比較三種二值化門檻

用 `for` 迴圈測試多組門檻，讓**同一張灰階模糊圖**產生三個不同結果。

**參考程式（先閱讀，再輸入下方空白 Code cell）：**

```python
plt.figure(figsize=(9, 3))
for i, value in enumerate([90, 127, 160]):
    _, result = cv2.threshold(smooth, value, 255, cv2.THRESH_BINARY)
    plt.subplot(1, 3, i + 1)
    plt.imshow(result, cmap="gray", vmin=0, vmax=255)
    plt.title(str(value))
    plt.axis("off")
plt.show()
```

**輪到你：** 將門檻清單改成 `[50, 100, 200]`，找出最能保留杯盤輪廓的一張。

## 第三節｜Canny、HSV 與 YOLO 物件偵測入門

先用 **Canny** 找亮暗變化，再用 **HSV** 選顏色區域。最後才認識 YOLO：**OpenCV 影像處理**不等於**YOLO 物件偵測**。物件偵測除了判斷類別，還要提供物件的位置和邊界框。

本節練習 **3-7 是選做延伸**，需要網路下載 YOLO11 預訓練模型。若無法連線，可以直接跳到 3-8，完成 YOLO 座標標註概念。

### 練習 3-1｜用 Canny 找貓咪照片的邊緣

先讀取 `cat.jpg` 並轉灰階，再使用 `cv2.Canny()`。**Canny 找的是亮暗變化**，不會直接知道線條代表耳朵、眼睛或鬍鬚。

**參考程式（先閱讀，再輸入下方空白 Code cell）：**

```python
cat = cv2.imread("cat.jpg")
cat_gray = cv2.cvtColor(cat, cv2.COLOR_BGR2GRAY)
edge = cv2.Canny(cat_gray, 60, 120)
plt.imshow(edge, cmap="gray", vmin=0, vmax=255)
plt.axis("off")
plt.show()
```

**輪到你：** 把門檻改成 `(30, 60)`，觀察邊緣線條是否變得比較多。

### 練習 3-2｜先模糊，再做 Canny

先模糊可能減少細碎毛髮的干擾，但也可能失去弱小線條。此處比較**直接偵測**與**先模糊再偵測**。

**參考程式（先閱讀，再輸入下方空白 Code cell）：**

```python
smooth_cat = cv2.GaussianBlur(cat_gray, (7, 7), 0)
edge2 = cv2.Canny(smooth_cat, 60, 140)
plt.figure(figsize=(9, 3))
plt.subplot(1, 2, 1)
plt.imshow(edge, cmap="gray")
plt.title("直接 Canny")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.imshow(edge2, cmap="gray")
plt.title("模糊後 Canny")
plt.axis("off")
plt.show()
```

**輪到你：** 將 `(7, 7)` 改為 `(15, 15)`，觀察細線是否變少。

### 練習 3-3｜HSV：把顏色拆成 H、S、V

HSV 分別代表**色相 H、飽和度 S、明度 V**。在 OpenCV 的 8-bit 影像中，H 通常是 **0～179**，S 與 V 為 **0～255**。這裡先看色相。

**參考程式（先閱讀，再輸入下方空白 Code cell）：**

```python
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
print(hsv.shape)
print("某個像素的 HSV：", hsv[120, 240])
plt.imshow(hsv[:, :, 0], cmap="hsv", vmin=0, vmax=179)
plt.title("H 色相")
plt.axis("off")
plt.show()
```

**輪到你：** 把 `hsv[:, :, 0]` 改成 `hsv[:, :, 2]`，觀察明度 V 的分布。

### 練習 3-4｜inRange：製作紅色範圍遮罩

`cv2.inRange()` 將符合 HSV 範圍的位置標成 255（白），不符合的標成 0（黑）。本範例使用投影片中的**較窄紅色區間**；不同照片的光線可能使結果不同。

**參考程式（先閱讀，再輸入下方空白 Code cell）：**

```python
mask = cv2.inRange(hsv, (0, 105, 45), (8, 255, 255))
print("遮罩尺寸：", mask.shape)
plt.imshow(mask, cmap="gray", vmin=0, vmax=255)
plt.axis("off")
plt.show()
```

**輪到你：** 將色相上限從 `8` 改為 `12`，比較白色區域的變化。

> 補充：紅色色相也可能出現在 OpenCV 色相刻度接近 179 的位置；此處先練習單一範圍，並非完整紅色偵測器。

### 練習 3-5｜bitwise_and：只保留選定的顏色

遮罩的**白色位置**保留原圖像素，**黑色位置**不顯示。注意黑色區域不代表原圖真的沒有物件。

**參考程式（先閱讀，再輸入下方空白 Code cell）：**

```python
only_red = cv2.bitwise_and(img, img, mask=mask)
plt.imshow(cv2.cvtColor(only_red, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

**輪到你：** 調整上一題的 HSV 上限後，重新執行本題，比較保留下來的杯盤範圍。

### 練習 3-6｜比較 Canny 和顏色遮罩

Canny 找**亮暗邊界**，HSV 遮罩找**符合顏色條件**的像素；兩者回答的是不同問題。

**參考程式（先閱讀，再輸入下方空白 Code cell）：**

```python
photo_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
photo_edge = cv2.Canny(photo_gray, 60, 120)
plt.figure(figsize=(9, 3))
plt.subplot(1, 2, 1)
plt.imshow(photo_edge, cmap="gray")
plt.title("Canny 邊緣")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.imshow(mask, cmap="gray")
plt.title("紅色遮罩")
plt.axis("off")
plt.show()
```

**輪到你：** 說明兩張黑白影像代表的意思，並試著改變 Canny 門檻。

### 練習 3-7｜選做｜用 YOLO11n 偵測照片中的物件

YOLO 會嘗試輸出**類別、方框與信心分數**，和 Canny 的邊緣圖不同。使用投影片對應的 `yolo11n.pt` **預訓練權重**。第一次使用需連網，可能會下載權重；模型不一定能辨認照片中的每個物件。若缺少 `ultralytics`，先在 Code cell 執行 `%pip install ultralytics`，再輸入下面的範例。**本題可跳過，不影響 3-8。**

**參考程式（先閱讀，再輸入下方空白 Code cell）：**

```python
from ultralytics import YOLO

model = YOLO("yolo11n.pt")
result = model.predict("cat.jpg", verbose=False)[0]
print("偵測到", len(result.boxes), "個物件")
view = result.plot()
plt.imshow(cv2.cvtColor(view, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

**輪到你：** 試著把 `"cat.jpg"` 改成 `"photo.jpg"`，比較兩張照片的偵測結果。不要把模型畫出的框當成人工標註。

> 網路、套件安裝或模型下載失敗時，請略過本題，先完成其他基本練習。

### 練習 3-8｜YOLO 的標註：將像素座標轉成 0～1

人工標註框不是推論結果。YOLO 偵測資料常將每個物件寫成 **`class x_center y_center width height`**；後四個數字要除以圖片寬、高。以下使用投影片的 **600×400** 座標練習，假設物件左上角是 `(120, 80)`、右下角是 `(420, 280)`，類別編號是 `0`。

**參考程式（先閱讀，再輸入下方空白 Code cell）：**

```python
width, height = 600, 400
x1, y1 = 120, 80
x2, y2 = 420, 280
x = ((x1 + x2) / 2) / width
y = ((y1 + y2) / 2) / height
w = (x2 - x1) / width
h = (y2 - y1) / height
line = f"0 {x:.4f} {y:.4f} {w:.4f} {h:.4f}"
print(line)
```

**輪到你：** 將方框改成左上角 `(60, 40)`、右下角 `(240, 200)`，再計算一次五個數字。檢查後四個數字是否都介於 0～1。

> 預期原範例結果：`0 0.4500 0.4500 0.5000 0.5000`。第一個 `0` 是類別編號，不是正規化座標。

## 補充｜YOLO 標註資料要如何整理？

投影片另有提供 HTML 標註工具，可以匯入圖片、畫 Bounding Box、設定類別，並匯出**每張圖片對應的 `.txt`** 及 `data.yaml`。

一個常見的資料集目錄是：

```text
dataset/
├── images/
│   ├── train/
│   └── val/
├── labels/
│   ├── train/
│   └── val/
└── data.yaml
```

- `images/train` 放訓練圖片；`images/val` 放驗證圖片。
- `labels/train`、`labels/val` 放與圖片**同檔名**的 YOLO `.txt` 標註檔。
- `data.yaml` 告訴模型資料集位置、類別數及類別名稱。
- **標註完成不代表模型已學會新類別。** 自訂類別通常仍需要另外訓練或微調；本週先理解觀念，不要求高一學生實作訓練。
