### 練習 3-1｜用 list 理解對話紀錄

### 練習 3-3｜用 previous_interaction_id 延續另一種對話

🔌 **需要網路與有效 API Key**

Interactions API 也能將前一次互動的 `id` 交給下一次呼叫，由服務端帶入對話脈絡。這是另一種**真實 API 對話**方法。

**參考程式（閱讀後，自己在下一個空白 Code cell 輸入）：**

```python
first = client.interactions.create(
    model="gemini-3.8-flash",
    input="我正在學 Python 的 dict。"
)
second = client.interactions.create(
    model="gemini-3.8-flash",
    input="請舉一個簡單例子。",
    previous_interaction_id=first.id
)
print(second.output_text)
```

**輪到你：** 觀察第二次回答是否仍在解釋 dict，並與「只有最新一句」的情況比較。

> 此模式可能在服務端保留互動紀錄；不要傳入需要保密的個人資訊。

### 練習 3-7｜用假設數字計算 Token 費用

💻 **不需要連線、不會呼叫 API**

輸入與輸出 Token 可能分別計價。下面費率**只是教學假設，並非 Google 官方報價**。費用公式是「Token 數 ÷ 100 萬 × 假設單價」。

**參考程式（閱讀後，自己在下一個空白 Code cell 輸入）：**

```python
input_tokens = 1200
output_tokens = 800
input_price = 0.20
output_price = 1.50
cost = input_tokens / 1_000_000 * input_price
cost += output_tokens / 1_000_000 * output_price
print("假設費用 USD：", round(cost, 6))
```

**輪到你：** 把兩種 Token 數都改成兩倍，觀察假設費用如何變化。
